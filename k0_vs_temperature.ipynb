```python
import matplotlib.pyplot as plt
import numpy as np
from ESM_ACDC_Forcing import rate_constants
```

```python
T_values = np.linspace(0,30,100)
T_base = 273.15

k_0,_,_ = rate_constants(T_values + T_base,S=35)

plt.figure(figsize=(8,5))
plt.plot(T_values,k_0,color='C0',linewidth=2.5)
plt.xlabel('Temperature [˚C]',fontsize=12)
plt.ylabel(r'$K_0$ [mol/kg/atm]',fontsize=12)
plt.title('CO$_2$ Solubility Constant vs. Temperature',fontsize=13,fontweight='bold')
plt.grid(alpha=0.3)
plt.gca().spines[['top','right']].set_visible(False)
plt.tight_layout()
```
